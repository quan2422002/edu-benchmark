```python
jsonl_file_path = "experiments/20260727_170150/outputs/benchmark_evaluation/judge_smoke_gemini35_blind_v2_retry1/run_judgments.jsonl"
import json

json_objects = []
with open(jsonl_file_path, "r", encoding="utf-8") as f:
    for idx, line in enumerate(f, 1):
        line = line.strip()
        if not line:
            continue
        try:
            json_objects.append(json.loads(line))
        except json.JSONDecodeError as e:
            print(f"Skipping invalid JSON on line {idx}: {e}")

len(json_objects)
```

```python
tmp_json_sample = None
for obj in json_objects:
    if obj.get("target_model_id") == "gemini-3.5-flash" and obj.get("benchmark_candidate_id") == "BC-HNMU-G9-R0224-STT13-AI08":
        tmp_json_sample = obj
        break
```
